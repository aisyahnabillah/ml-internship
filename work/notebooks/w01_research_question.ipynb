Two real numbers from the starter dataset:

2. Median `content_age_days` by `trend_direction`: declining pages have a median age of 216 days, while stable pages have a median age of 300 days, and up-trending pages sit at 291.5 days, a gap of roughly 75-84 days between declining and non-declining groups.
